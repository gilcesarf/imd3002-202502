```python
import pandas as pd
import time

## Carregando o dataset
dataset = pd.read_csv('HOG_128_30x30.csv', encoding='utf-8')

# Selecionando todos os atributos com exceção da classe
X = dataset.iloc[:, :-1]  # Features
y = dataset.classe        # Target variable (classe)
```

```python
## Carregando o Voting e os algoritmos base (DT), (MLP), (k-NN) e (NB)
from sklearn.ensemble import VotingClassifier
from sklearn.neural_network import MLPClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.tree import DecisionTreeClassifier
from sklearn import metrics

# Separando dataset em duas partes: treinamento e teste
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=19)
```

```python
## Definição das técnicas e seus parâmetros
clf1 = MLPClassifier(hidden_layer_sizes=(100,50), activation='relu',
                     solver='adam', max_iter=1500, learning_rate_init=0.001)

clf2 = MLPClassifier(hidden_layer_sizes=(70,70), activation='relu',
                     solver='adam', max_iter=2500, learning_rate_init=0.0001)

clf3 = MLPClassifier(hidden_layer_sizes=(120), activation='identity',
                     solver='sgd', max_iter=1500, learning_rate_init=0.01)

clf4 = KNeighborsClassifier(n_neighbors=7, metric='euclidean')

clf5 = KNeighborsClassifier(n_neighbors=9, metric='euclidean')

clf6 = KNeighborsClassifier(n_neighbors=10, metric='euclidean')

clf7 = GaussianNB(priors = None, var_smoothing=1e-09)

clf8 = DecisionTreeClassifier(criterion="gini", max_depth=2)

clf9 = DecisionTreeClassifier(criterion="gini", max_depth=5)

clf10 = DecisionTreeClassifier(criterion="entropy", max_depth=9)

start = time.time()

### Configurando o Voting
vt = VotingClassifier(estimators=[('MLP01', clf1), ('MLP02', clf2), ('MLP03', clf3),
 ('kNN01', clf4), ('kNN02', clf5), ('kNN03', clf6), ('NB01', clf7), ('DT01', clf8),
  ('DT02', clf9), ('DT03', clf10)], voting='soft')

vt.fit(X_train,y_train)
y_pred = vt.predict(X_test)

end_time = time.time()

## Model Accuracy
f1_score = metrics.f1_score(y_test, y_pred, average = "weighted")
print('F1_Score: %.3f' % f1_score)
print("Tempo de execução: ", end_time - start)
```
