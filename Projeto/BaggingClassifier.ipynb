# **Bagging**

```python
import pandas as pd

## Carregando o dataset
dataset = pd.read_csv('HOG_128_30x30.csv', encoding='utf-8')

## Selecionando todos os atributos com exceção da classe
X = dataset.iloc[:, :-1]  # Features
y = dataset.classe        # Target variable (classe)
```

```python
## Carregando o Bagging e os algoritmos base (DT), (MLP), (k-NN) e (NB)
from sklearn.ensemble import BaggingClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.neural_network import MLPClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn import metrics

# Separando dataset em duas partes: treinamento e teste
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=19)
```

# Bagging com DT

```python
## Instanciando Bagging com DecisionTree
bgclassifier = BaggingClassifier(estimator=DecisionTreeClassifier(criterion="gini", max_depth=2),
                                 n_estimators=10, max_features=1.0, max_samples=1.0)

bgclassifier.fit(X_train,y_train)
y_pred = bgclassifier.predict(X_test)

## Model Accuracy
f1_score = metrics.f1_score(y_test, y_pred, average = "weighted")
print('F1_Score: %.3f' % f1_score)
```

# Bagging com k-NN

```python
## Instanciando Bagging com KNeighbors
bgclassifier = BaggingClassifier(estimator=KNeighborsClassifier(n_neighbors=10, metric='euclidean'),
                                 n_estimators=10, max_features=1.0, max_samples=1.0)

bgclassifier.fit(X_train,y_train)
y_pred = bgclassifier.predict(X_test)

## Model Accuracy
f1_score = metrics.f1_score(y_test, y_pred, average = "weighted")
print('F1_Score: %.3f' % f1_score)
```

# Bagging com NB

```python
## Instanciando Bagging com GaussianNB
bgclassifier = BaggingClassifier(estimator=GaussianNB(priors = None, var_smoothing=1e-09),
                                  n_estimators=10, max_features=1.0, max_samples=1.0)

bgclassifier.fit(X_train,y_train)
y_pred = bgclassifier.predict(X_test)

## Model Accuracy
f1_score = metrics.f1_score(y_test, y_pred, average = "weighted")
print('F1_Score: %.3f' % f1_score)
```

# Bagging com MLP

```python
## Utilizando uma métrica de tempo
import time

start = time.time()
## Instanciando Bagging com MLPClassifier
bgclassifier = BaggingClassifier(estimator=MLPClassifier(hidden_layer_sizes=(100,50), activation='relu', solver='adam',
                                                        max_iter=1500, learning_rate_init=0.001),
                                                        n_estimators=10, max_features=1.0, max_samples=1.0)

bgclassifier.fit(X_train,y_train)
y_pred = bgclassifier.predict(X_test)

end_time = time.time()

## Model Accuracy
f1_score = metrics.f1_score(y_test, y_pred, average = "weighted")
print('F1_Score: %.3f' % f1_score)
print("Tempo de execução: ", end_time - start)
```
