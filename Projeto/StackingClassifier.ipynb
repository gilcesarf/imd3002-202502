```python
import pandas as pd
import time

## Carregando o dataset
dataset = pd.read_csv('HOG_128_30x30.csv', encoding='utf-8')

# Selecionando todos os atributos com exceção da classe
X = dataset.iloc[:, :-1]  # Features
y = dataset.classe        # Target variable (classe)
```

```python
## Carregando o StackingClassifier
from sklearn.ensemble import StackingClassifier
from sklearn.neural_network import MLPClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.tree import DecisionTreeClassifier

# Algoritmo para fazer a fusão
from sklearn.linear_model import LogisticRegression

from sklearn import metrics

# Separando dataset em duas partes: treinamento e teste
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=19)
```

```python
## Definição das técnicas e seus parâmetros
estimators = [('MLP01', MLPClassifier(hidden_layer_sizes=(100,50), activation='relu',
                      solver='adam', max_iter=1500, learning_rate_init=0.001)),
              ('MLP02', MLPClassifier(hidden_layer_sizes=(70,70), activation='relu',
                     solver='adam', max_iter=2500, learning_rate_init=0.0001)),
              ('MLP03', MLPClassifier(hidden_layer_sizes=(120), activation='identity',
                     solver='sgd', max_iter=1500, learning_rate_init=0.01)),
              ('kNN01', KNeighborsClassifier(n_neighbors=7, metric='euclidean')),
              ('kNN02', KNeighborsClassifier(n_neighbors=9, metric='euclidean')),
              ('kNN03', KNeighborsClassifier(n_neighbors=10, metric='euclidean')),
              ('DT01', DecisionTreeClassifier(criterion="gini", max_depth=2)),
              ('DT02', DecisionTreeClassifier(criterion="gini", max_depth=5)),
              ('DT03', DecisionTreeClassifier(criterion="entropy", max_depth=9)),
              ('NB01', GaussianNB(priors = None, var_smoothing=1e-09)) ]

start = time.time()

stk = StackingClassifier(estimators=estimators, final_estimator=LogisticRegression())

stk.fit(X_train,y_train)
y_pred = stk.predict(X_test)

end_time = time.time()

## Model Accuracy
f1_score = metrics.f1_score(y_test, y_pred, average = "weighted")
print('F1_Score: %.3f' % f1_score)
print("Tempo de execução: ", end_time - start)
```
