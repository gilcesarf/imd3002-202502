```python
import pandas as pd
from numpy import mean
from numpy import std

## Carregando o dataset
#dataset = pd.read_csv('HOG_128_20x20.csv', encoding='utf-8')
dataset = pd.read_csv('HOG_128_30x30.csv', encoding='utf-8')

## Selecionando todos os atributos com exceção da classe
X = dataset.iloc[:, :-1]  # Features
y = dataset.classe        # Target variable (classe)

## Carregando o algoritmo
from sklearn.neural_network import MLPClassifier
from sklearn import metrics
from sklearn.metrics import confusion_matrix

## Implementando k-fold CV
from sklearn.model_selection import cross_val_score
from sklearn.model_selection import cross_val_predict
from sklearn.model_selection import KFold

# 5-fold CV
kf = KFold(n_splits=5, random_state=1, shuffle=True)

### MLPClassifier e seus Hyperparametros
mlp = MLPClassifier(hidden_layer_sizes=(100), activation='relu', solver='adam', max_iter=1000)

# Model Accuracy
scores = cross_val_score(mlp, X, y, scoring='accuracy', cv=kf)
print('Accuracy: %.3f (%.3f)' % (mean(scores), std(scores)))

# Matriz de confusão p/ kf
y_pred = cross_val_predict(mlp, X, y, cv=kf)
confusion_matrix(y, y_pred)
```

```python
### GridSearch
from sklearn.model_selection import GridSearchCV

# Separando dataset em duas partes: treinamento e teste
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=19)

## Utilizando uma métrica de tempo
import time

# Definir o modelo
mlp = MLPClassifier(random_state=42)

# Definir os hiperparâmetros a serem testados
param_grid = {
    'hidden_layer_sizes': [(50), (100), (120), (150), (70,70), (100,50)],
    'activation': ['identity', 'logistic', 'tanh', 'relu'],
    'solver': ['sgd', 'adam'],
    'learning_rate_init': [0.0001, 0.001, 0.01, 0.1],
    'max_iter': [500, 1000, 1500, 2000]
}

# Configurar o GridSearchCV
grid_search = GridSearchCV(
    estimator=mlp,
    param_grid=param_grid,
    scoring='accuracy',  # Pode ser alterado para outra métrica (ex.: 'f1_weighted', 'roc_auc')
    cv=3,                # Validação cruzada com 3 folds
    verbose=2,           # Para exibir detalhes durante a execução
    n_jobs=-1            # Usa todos os núcleos disponíveis
)

start_time = time.time()

# Executar a busca
grid_search.fit(X_train,y_train)

# Melhor combinação de parâmetros
print("Melhores parâmetros encontrados:", grid_search.best_params_)

# Melhor pontuação obtida
print("Melhor pontuação:", grid_search.best_score_)

end_time = time.time()
# Tempo total de treinamento
execution_time = end_time - start_time
print(f"Tempo de execução para treinamento: {execution_time:.4f} segundos")
```
